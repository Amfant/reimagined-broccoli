### What parameters are most useful in better reproducing the sizes of each of these planets? Is it realistic?

The most useful parameters for matching planetary radii in ExoPlex are:

Fe/Mg ratio: to control the core size and overall density

Si/Mg ratio: to control the mantle mineralogy and volume
Mantle FeO content: which shifts Fe from the dense core to the mantle and expands the radius

Light elements in the core: to simulate buoyancy without increasing mass

Using solar composition values (FeMg=0.9, SiMg=0.9), ExoPlex closely reproduces Earth’s radius — which is expected and realistic.

To match Mars, you often need a slightly more silicate-rich mantle and more Fe in the mantle (FeO), which is still within reason.

To match Mercury, however, you usually need an unrealistically high Fe/Mg ratio (1.5+), which is only physically plausible if Mercury lost a large part of its mantle during formation or had an unusual disk composition — so it’s less realistic unless you include such processes.

## Logging the Radius Results

Using mass and radius values from Agol et al. (2021), we modeled TRAPPIST-1e in ExoPlex with FeMg = 0.75 and SiMg = 1.02, reflecting the estimated host star abundance ratios. The input mass was 0.772 M⊕, and the modeled radius was 0.949 R⊕, in excellent agreement with the observed value of 0.920 ± 0.013 R⊕. No atmosphere was required to explain the observed size. The model predicts a core mass fraction of 22.7%, lower than Earth’s (~32%), consistent with a less iron-rich interior. Mineralogical output indicates a silicate-dominated mantle with moderate garnet and perovskite fractions at depth, similar to Earth but with potentially more oxidized mantle due to lower FeMg and added FeO.

## Final Structure Range (Answer to Q4)

Based on ExoPlex models with adjusted FeMg (0.65–0.75), SiMg (1.02), and mantle FeO (up to 0.05), TRAPPIST-1e likely has a core mass fraction of 20–25%. Light elements in the core (O, S) can raise the radius without increasing mass. The modeled radius of 0.89–0.95 R⊕ brackets the observed value (0.920 ± 0.013 R⊕). If a thinner model is required, a thin atmosphere layer (~100–200 km) may explain the excess radius without contradicting a rocky interior.

##Final Comparison (Answer to Q5)

The mantle mineralogy of TRAPPIST-1e is broadly similar to Earth’s, dominated by silicates such as olivine, garnet, and perovskite. However, due to its lower FeMg and added FeO, TRAPPIST-1e may have more oxidized minerals, potentially increasing garnet and decreasing olivine. At higher pressures, the model shows increased perovskite and akimotoite, consistent with a compact rocky structure. Overall, mineralogy supports a rocky, silicate-rich planet comparable to Earth, but with slightly different core/mantle distribution.
